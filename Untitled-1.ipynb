```python
import numpy as np
import pandas as pd  
import matplotlib.pyplot as plt 
import seaborn as sns
import random
import tensorflow as tf 
```

```python
seed = 167
np.random.seed(seed)
random.seed(seed)
tf.random.set_seed(seed)
```

```python
df = pd.read_csv(r"E:\belajar\archive_18\pakwheels_used_cars_2026_clean.csv")
df
```

```python
df = df.drop(df[df['price_outlier_flag'] == "True"].index)
```

```python
korelasi = df[['year','price_pkr','mileage_km','engine_cc']].corr()
korelasi
```

```python
df = df[['year','price_pkr','mileage_km','engine_cc']]
df.isna().sum()
```

```python
df = df.dropna()
```

```python
x = df
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
xscaler = scaler.fit_transform(x)
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

inertia, silhouette = [], []
for k in range (1, 11):
    model_k = KMeans(n_clusters = k, n_init = 10).fit(xscaler)
    model_k.fit
    inertia.append(model_k.inertia_)
    if k >= 2 :
        silhouette.append(silhouette_score(xscaler, model_k.labels_))
fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ax[0].plot(range(1, 11), inertia, marker='o'); ax[0].set_title('Elbow (inertia)')
ax[1].plot(range(2, 11), silhouette, marker='o'); ax[1].set_title('Silhouette')
plt.show()
```

```python
from sklearn.cluster import DBSCAN
from sklearn.neighbors import NearestNeighbors

nn = NearestNeighbors(n_neighbors=5).fit(xscaler)
jarak, _ = nn.kneighbors(xscaler)
plt.plot(np.sort(jarak[:, -1]))
plt.xlabel('Titik (urut)'); plt.ylabel(f'Jarak ke tetangga ke-5'); plt.show()
```

```python
eps_values = [0.3, 0.4, 0.5, 0.6, 0.7, 0.8, 0.9, 1.0]
silhouette_scores = []
valid_eps = []

# 2. Looping untuk menghitung silhouette score pada setiap eps
for eps in eps_values:
    # Menggunakan min_samples=5 sesuai dengan grafik tetangga ke-5 Anda
    dbscan = DBSCAN(eps=eps, min_samples=5).fit(xscaler)
    labels = dbscan.labels_
    
    # Filter dan abaikan noise (-1) agar kalkulasi score akurat
    non_noise_mask = labels != -1
    unique_labels = set(labels[non_noise_mask])
    
    # Silhouette score hanya bisa berjalan jika terbentuk minimal 2 klaster
    if len(unique_labels) > 1:
        score = silhouette_score(xscaler[non_noise_mask], labels[non_noise_mask])
        silhouette_scores.append(score)
        valid_eps.append(eps)
        print(f"eps: {eps:.1f} -> Silhouette Score (Tanpa Noise): {score:.4f}")
    else:
        print(f"eps: {eps:.1f} -> Gagal menghitung skor (jumlah klaster < 2 atau semua data dianggap noise)")

# 3. Membuat Plot Menggunakan plt
plt.figure(figsize=(7, 4))
plt.plot(valid_eps, silhouette_scores, marker='o', linestyle='-', color='b')
plt.title('Silhouette Score untuk Berbagai Nilai EPS pada DBSCAN')
plt.xlabel('Nilai EPS')
plt.ylabel('Silhouette Score')
plt.grid(True)
plt.show()
```

```python
for eps in [0.7, 0.8, 0.9, 1.0]:
    dbscan = DBSCAN(eps=eps, min_samples=5).fit(xscaler)
    labels = dbscan.labels_
    
    n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
    n_noise = list(labels).count(-1)
    pct_noise = (n_noise / len(labels)) * 100
    
    print(f"eps: {eps:.1f} -> Jumlah Klaster: {n_clusters} | Total Noise: {n_noise} ({pct_noise:.2f}%)")
```

```python
#cek balance
y = dbscan.labels_
class_counts = pd.Series(y).value_counts()
class_counts
```

```python
from sklearn.model_selection import train_test_split
y = dbscan.labels_
x_train, x_test, y_train, y_test = train_test_split(xscaler,y, test_size = 0.2, stratify= y, random_state= seed)
```

```python
from imblearn.under_sampling import RandomUnderSampler
rus = RandomUnderSampler()
x_rus, y_rus = rus.fit_resample(x_train,y_train)
```

```python
from sklearn.svm import SVC
svm_rus = SVC()
svm_rus.fit(x_rus,y_rus)
```

```python
y_pred_svmrus = svm_rus.predict(x_test)
```

```python
svm = SVC()
svm.fit(x_train,y_train)
```

```python
y_pred_svm = svm.predict(x_test)
```

```python
from sklearn.tree import DecisionTreeClassifier
dtc_rus = DecisionTreeClassifier()
dtc_rus.fit(x_rus, y_rus)
y_pred_dtcrus = dtc_rus.predict(x_test)
```

```python
dtc = DecisionTreeClassifier()
dtc.fit(x_train, y_train)
y_pred_dtc = dtc.predict(x_test)
```

```python
from sklearn.naive_bayes import GaussianNB
nvb_rus = GaussianNB()
nvb_rus.fit(x_rus,y_rus)
y_pred_nvbrus = nvb_rus.predict(x_test)
```

```python
nvb = GaussianNB()
nvb.fit(x_train, y_train)
y_pred_nvb = nvb.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score, precision_recall_curve, precision_score, recall_score, f1_score, ConfusionMatrixDisplay, confusion_matrix 

print("accuracy svm tanpa balanced:",accuracy_score(y_test,y_pred_svm))
print("accuracy svm dengan rus:",accuracy_score(y_test, y_pred_svmrus))
print("presisi svm baseline:", precision_score(y_test,y_pred_svm, average = 'macro'))
print("presisi svm dengan rus:", precision_score(y_test, y_pred_svmrus, average = 'macro'))
print("recall svm baseline:", recall_score(y_test, y_pred_svm, average = 'macro'))
print("recall svm dengan rus:", recall_score(y_test, y_pred_svmrus, average = 'macro'))
print("f1 score svm baseline:", f1_score(y_test, y_pred_svm, average = 'macro'))
print("f1 score svm dengan rus:", f1_score(y_test, y_pred_svmrus, average = 'macro'))
```

```python
print("accuracy dtc tanpa balanced:",accuracy_score(y_test,y_pred_dtc))
print("accuracy dtc dengan rus:",accuracy_score(y_test, y_pred_dtcrus))
print("presisi dtc baseline:", precision_score(y_test,y_pred_dtc, average = 'macro'))
print("presisi dtc dengan rus:", precision_score(y_test, y_pred_dtcrus, average = 'macro'))
print("recall dtc baseline:", recall_score(y_test, y_pred_dtc, average = 'macro'))
print("recall dtc dengan rus:", recall_score(y_test, y_pred_dtcrus, average = 'macro'))
print("f1 score dtc baseline:", f1_score(y_test, y_pred_dtc, average = 'macro'))
print("f1 score dtc dengan rus:", f1_score(y_test, y_pred_dtcrus, average = 'macro'))
```

```python
print("accuracy nvb tanpa balanced:",accuracy_score(y_test,y_pred_nvb))
print("accuracy nvb dengan rus:",accuracy_score(y_test, y_pred_nvbrus))
print("presisi nvb baseline:", precision_score(y_test,y_pred_nvb, average = 'macro'))
print("presisi nvb dengan rus:", precision_score(y_test, y_pred_nvbrus, average = 'macro'))
print("recall nvb baseline:", recall_score(y_test, y_pred_nvb, average = 'macro'))
print("recall nvb dengan rus:", recall_score(y_test, y_pred_nvbrus, average = 'macro'))
print("f1 score nvb baseline:", f1_score(y_test, y_pred_nvb, average = 'macro'))
print("f1 score nvb dengan rus:", f1_score(y_test, y_pred_nvbrus, average = 'macro'))
```

```python
fig, ax = plt.subplots(3, 2, figsize=(10, 14))

hasil = [
    ("SVM",           y_pred_svm, y_pred_svmrus),
    ("Decision Tree", y_pred_dtc, y_pred_dtcrus),
    ("Naive Bayes",   y_pred_nvb, y_pred_nvbrus),
]

for i, (nama, base, rus_pred) in enumerate(hasil):
    ConfusionMatrixDisplay.from_predictions(y_test, base, ax=ax[i, 0])
    ax[i, 0].set_title(f"{nama} - Baseline")
    ConfusionMatrixDisplay.from_predictions(y_test, rus_pred, ax=ax[i, 1])
    ax[i, 1].set_title(f"{nama} - Undersampling")

plt.tight_layout()
fig.savefig('confusion_matrix.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
import pickle

bundle = {
    "features": ["year", "price_pkr", "mileage_km", "engine_cc"],
    "scaler": scaler,
    "models": {
        "SVM - Baseline": svm,
        "SVM - Undersampling": svm_rus,
        "Decision Tree - Baseline": dtc,
        "Decision Tree - Undersampling": dtc_rus,
        "Naive Bayes - Baseline": nvb,
        "Naive Bayes - Undersampling": nvb_rus,
    },
    "data": df.assign(cluster=dbscan.labels_),
}

with open("model_bundle.pkl", "wb") as f:
    pickle.dump(bundle, f)
```
